```python
import pandas as pd
import os
import numpy as np
from datetime import datetime
from zoneinfo import ZoneInfo
import warnings
warnings.filterwarnings("ignore", category=UserWarning, module="openpyxl")
pd.set_option('display.max_columns', None)
```

```python
df_desgravamen= pd.read_excel('C:/DATOS/TRAMAS DE ERRORES PRESTAMOS - Enero- Junio26 - muestra.xlsx', sheet_name=0, 
                              dtype={'DESCRIPCION ERROR': str, 'CÓD. CERTIFICADO CANAL': str, 'CÓDIGO ERROR': str })
df_desgravamen.columns = (df_desgravamen.columns.str.strip().str.upper()
                            .str.replace(r'[^A-Za-z0-9]', '_', regex=True))
```

```python
df_desgravamen= df_desgravamen[['C_D__CERTIFICADO_CANAL','TIPO_MOVIMIENTO','C_DIGO_ERROR', 'DESCRIPCI_N_ERROR',
                                'MONEDA','PRIMABRUTACAN','FEC__INICIO','FEC__FIN']]
```

```python
df_desgravamen= df_desgravamen.rename(columns={'C_D__CERTIFICADO_CANAL':'CODIGO_CERTIFICADO',
                                               'C_DIGO_ERROR':'CODIGO_ERROR', 'DESCRIPCI_N_ERROR':'DESCRIPCION_ERROR',                                            
                                                'FEC__INICIO':'FECHA_INICIO', 'FEC__FIN':'FECHA_FIN', 
                                                'PRIMABRUTACAN':'PRIMA_BRUTA'})
```

```python

filtro = df_desgravamen['CODIGO_CERTIFICADO'].str.startswith('C', na=False)
df_desgravamen.loc[filtro, 'CODIGO_CERTIFICADO'] = (df_desgravamen.loc[filtro, 'CODIGO_CERTIFICADO'].str[1:])
df_desgravamen['FECHA_INICIO']= pd.to_datetime(df_desgravamen['FECHA_INICIO'], errors='coerce').dt.date
df_desgravamen['FECHA_FIN']= pd.to_datetime(df_desgravamen['FECHA_FIN'], errors='coerce').dt.date
df_desgravamen['CODIGO_ERROR'] = df_desgravamen['CODIGO_ERROR'].apply(lambda x: str(x).zfill(4) if pd.notnull(x) else "")
```
