```python
import pandas as pd
import os
import numpy as np
from datetime import datetime
from zoneinfo import ZoneInfo
import warnings
warnings.filterwarnings("ignore", category=UserWarning, module="openpyxl")
pd.set_option('display.max_columns', None)
```

```python
df_resumen_reclamos= pd.read_excel('C:/DATOS/Resumen Devoluciones de Primas FEB 26.xlsx', sheet_name='RECLAMOS',
                              dtype={'AÑO': str, 'N° DE OPERACIoN': str, 'RESPONSABLE':str, 'COMENTARIO':str})
df_resumen_reclamos.columns = (df_resumen_reclamos.columns.str.strip().str.upper()
                            .str.replace(r'[^A-Za-z0-9]', '_', regex=True))
```

```python
df_resumen_reclamos= df_resumen_reclamos.rename(columns={'A_O':'ANIO','N__DE_OPERACION':'NRO_OPERACION'})
```

```python
df_resumen_reclamos.head(3)
```

```python
df_resumen_reclamos['PRODUCTO'] = df_resumen_reclamos["PRODUCTO"].str.upper()
df_resumen_reclamos['FECHA_OPERACION']= pd.to_datetime(df_resumen_reclamos['FECHA_OPERACION'], errors='coerce').dt.date
df_resumen_reclamos['FECHA_CARGA'] = datetime.now(ZoneInfo("America/Lima")).strftime("%Y-%m-%d")
df_resumen_reclamos['FECHA_CARGA'] = pd.to_datetime(df_resumen_reclamos['FECHA_CARGA'], errors='coerce').dt.date
```
